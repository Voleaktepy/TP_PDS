```python
#step1 create one student as a tuple and print the name
student=("Dara",20,3.45)
print("Name:",student[0])
```

```python
#step2 unpack the tuple:put its three values into their variables in one line
name,age,gpa=student
print(name,"is",age,"years old","GPA",gpa) 
```

```python
#Step3 Make a list of four student tuples and print each one with a for loop that unpacks the tuple
student=[("Data",20,3.45),("Sophea",21,3.92),("Vibol",19,3.10),("Malis",22,3.78)]
for name , age , gpa in student:
    print(name,age,gpa)
```

```python
#Step4 Sort the student by GPA,best first.key=lambda s:s[2]tells sorted()to compare the item at index2(the GPA).The best student is the first tuple of the result
ranking=sorted(student,key=lambda s:s[2],reverse=True)
print("Best student:",ranking[0][0])
```

```python
#Step5 Try to change the GPA inside the tuple. Python refuses;try/expect catches the error so the program can continue
try:
    student[2]=4.0
except TypeError as error:
    print("Error:",error)    
```

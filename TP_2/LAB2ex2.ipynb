```python
#step1
shopping=["rice","egg","milk"]
print("Start: ",shopping)
```

```python
#step2
shopping.append("bread")
shopping.insert(0,"water")
print("Added : ",shopping)
```

```python
#step3
shopping.remove("egg")
last=shopping.pop()
print("Remove :",last)
print("Now :",shopping)
                
```

```python
#step4
print("First 2:",shopping[:2])
print("Last 2",shopping[-2:])
print("Reverse",shopping[::-2])
```

```python
#step5
print("Is milk in the last?","milk" in shopping)
for i , item in enumerate(shopping,start=1):
    print(i, item)
```

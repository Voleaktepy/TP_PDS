```python
#Step 1. The football list has a duplicate ("Dara" registered twice). Turn it into a set and compare the sizes.
football_list = ["Dara", "Vibol", "Kosal", "Dara", "Bopha"]
football = set(football_list)
print("List size:", len(football_list), "| Set size:", len(football))
```

```python
#Step 2. Create the music club directly as a set with curly braces.
music = {"Sophea", "Kosal", "Malis", "Dara"}
```

```python
#Step 3. Use the four set operators. A set has no order, so wrap each result in sorted() to print it alphabetically.

print("Both clubs    :", sorted(football & music))   # intersection
print("Any club      :", sorted(football | music))   # union
print("Football only :", sorted(football - music))   # difference
print("Exactly one   :", sorted(football ^ music))   # symmetric difference
```

```python
#Step 4. Add "Chenda" to the music club and remove "Malis". Then check whether "Vibol" is in the music club.

music.add("Chenda")
music.discard("Malis")
print("Music now     :", sorted(music))
print("Is Vibol in music?", "Vibol" in music)
```
